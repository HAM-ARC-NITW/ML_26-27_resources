```python
import numpy as np
```

```python
#load the dataset
data = np.genfromtxt('cricket_data.csv', delimiter=",", skip_header=1)
balls_faced = data[:, 0]
runs_scored = data[:, 1]
```

```python
X = balls_faced.reshape(-1, 1)
y = runs_scored
```

```python
print("first few rows of the dataset:")
for i in range(5):
    print(balls_faced[i], runs_scored[i])
print()
```

```python
# gradient descent implementation from scratch
def train_gradient_descent(x, y, lr=0.01, epochs=5000):
    # scale x first, otherwise gradient descent blows up because
    # balls_faced ranges from 10 to 70 and the updates overshoot
    x_mean = x.mean()
    x_std = x.std()
    x_scaled = (x - x_mean) / x_std

    m = 0.0
    c = 0.0
    n = len(x)
    losses = []

    for i in range(epochs):
        y_pred = m * x_scaled + c
        error = y_pred - y

        dm = (2 / n) * np.sum(error * x_scaled)
        dc = (2 / n) * np.sum(error)

        m = m - lr * dm
        c = c - lr * dc

        losses.append(np.mean(error ** 2))

    # convert m and c back to the original unscaled x
    m_final = m / x_std
    c_final = c - m_final * x_mean

    return m_final, c_final, losses
```

```python
#do da training
m, c, loss_history = train_gradient_descent(balls_faced, runs_scored)
```

```python
print("gradient descent result:")
print("slope =", round(m, 4))
print("intercept =", round(c, 4))
print("final loss =", round(loss_history[-1], 3))
```

```python
# now check it against sklearn's built in linear regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()
model.fit(X, y)
predictions = model.predict(X)
```

```python
print("sklearn result:")
print("slope =", round(model.coef_[0], 4))
print("intercept =", round(model.intercept_, 4))
print("mse =", round(mean_squared_error(y, predictions), 3))
print("r2 =", round(r2_score(y, predictions), 4))
```

```python
# predict for a new value, say a batsman who faced 45 balls
new_balls = 45
pred1 = m * new_balls + c
pred2 = model.predict([[new_balls]])[0]
print("prediction for 45 balls faced:")
print("from scratch:", round(pred1, 1))
print("sklearn:", round(pred2, 1))
```

```python
# plotting
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 2, figsize=(12, 5))
x_line = np.linspace(balls_faced.min(), balls_faced.max(), 100)
y_line = m * x_line + c

axs[0].scatter(balls_faced, runs_scored, label="actual data")
axs[0].plot(x_line, y_line, color="red", label="fitted line")
axs[0].set_xlabel("Balls Faced")
axs[0].set_ylabel("Runs Scored")
axs[0].set_title("Runs Scored vs Balls Faced")
axs[0].legend()

axs[1].plot(loss_history)
axs[1].set_xlabel("epoch")
axs[1].set_ylabel("mean squared error")
axs[1].set_title("loss during training")
```
