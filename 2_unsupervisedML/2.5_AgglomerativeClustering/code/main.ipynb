```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
# Create a blob of 200 data points
dataset = make_blobs(n_samples = 200,
                n_features = 2,
                centers = 4,
                cluster_std = 1.6,
                random_state = 50
               )
```

```python
points = dataset[0]
```

```python
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering
```

```python
# create a dendogram
dendogram = sch.dendrogram(sch.linkage(points, method='ward'))
```

```python
plt.scatter(dataset[0][:, 0], dataset[0][:, 1])

```

```python
# performing agglomerrative clustering
hc = AgglomerativeClustering(n_clusters=4, linkage='ward')
```

```python
y_hc = hc.fit_predict(points)

# all data pts in cluster one
plt.scatter(points[y_hc == 0, 0], points[y_hc == 0,1], s=10, c='green')

# all data pts cluster two
plt.scatter(points[y_hc == 1, 0], points[y_hc == 1,1], s=10, c='red')

# all data pts cluster two
plt.scatter(points[y_hc == 2, 0], points[y_hc == 2,1], s=10, c='cyan')

# all data pts cluster two
plt.scatter(points[y_hc == 3, 0], points[y_hc == 3,1], s=10, c='purple')
```
